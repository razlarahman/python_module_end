```python
Q1 = df["engagement_rate"].quantile(0.25)
print(Q1)
```

```python
Q3 = df["engagement_rate"].quantile(0.75)
print(Q3)
```

```python
IQR = Q3 - Q1
print(IQR)
```

```python
lower_limit = Q1 - 1.5 * IQR
print(lower_limit)
```

```python
upper_limit = Q3 + 1.5 * IQR
print(upper_limit)
```

```python
df[
    (df["engagement_rate"] < lower_limit) |
    (df["engagement_rate"] > upper_limit)
]
```

```python
median_engagement_rate = df["engagement_rate"].median()

df.loc[
    (df["engagement_rate"] < lower_limit) |
    (df["engagement_rate"] > upper_limit),
    "engagement_rate"
]=median_engagement_rate
```

```python
#task6 data visualisation
import matplotlib.pyplot as plt
plt.figure(figsize=(8, 6), dpi=300)
plt.scatter(df["likes"],df["impression_count"],s=10, alpha=0.8)
plt.xlabel("likes")
plt.ylabel("impression_count")
plt.title("likes vs impression_count")
```

```python
plt.plot(df["posted_at"],df["engagement_rate"],color="red",marker="o")
plt.xlabel("posted_at")
plt.ylabel("engagement_rate")
plt.xticks(rotation=45)
plt.title("posted_at vs engagement_rate")
plt.show()
```

```python
count=pd.crosstab(df["post_category"], df["post_type"])

count.plot(kind="bar", stacked=True, color=["red", "blue","yellow","green"], width=0.6)
plt.xlabel("post_category")
plt.ylabel("count")
plt.title("post_category vs post_type")
plt.show()
```
